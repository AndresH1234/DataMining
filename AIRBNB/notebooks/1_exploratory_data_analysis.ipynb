## Gráficas

```python
def outliers(s):
    # Ver los que están por encima del Q3 + 1.5 * IQR y Q1 - 1.5 * IQR
    Q1 = df[s].quantile(0.25)
    Q3 = df[s].quantile(0.75)
    IQR = Q3 - Q1
    print(df[(df[s] < Q1 - 1.5 * IQR) | (df[s] > Q3 + 1.5 * IQR)])
```

```python
# Función para ver los outliers de una columna con Z-Score
def outliers_zscore(s):
    threshold = 3
    mean_y = df[s].mean()
    stdev_y = df[s].std()
    z_scores = [(y - mean_y) / stdev_y for y in df[s]]
    print(df[abs(z_scores) > threshold])
```
